```python
#Create Unique List Without set() 
numbers = [10, 20, 10, 30, 20, 40, 30, 50]
unique_numbers = []
for num in numbers:
    if num not in unique_numbers:
        unique_numbers.append(num)
print("Unique number list", unique_numbers)
print("Orginal list", numbers)
print("Unique number using set()", list(set(numbers)))
```
