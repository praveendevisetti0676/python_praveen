```python
#Count Positive, Negative and Zero Values 
numbers = [10, -4, 8, -2, 0, 15, -9, 21]
positive_count = 0
negative_count = 0  
number_of_zeros = 0
for num in numbers:
    if num > 0:
        positive_count += 1
    elif num < 0:
        negative_count += 1
    else:
        number_of_zeros += 1

print("Positive numbers:", positive_count)
print("Negative numbers:", negative_count)
print("Zeros:", number_of_zeros)
```
