```python
transactions = [1200, 450, 800, 1500, 2300, 700, 100]
# Find the highest and lowest transaction value without using max() and min()
highest = transactions[0]
lowest = transactions[0]
for i in transactions:
    if i > highest: 
        highest = i 
    if i < lowest: 
        lowest = i 
print("Highest transaction value: ₹", highest)
print("Lowest transaction value: ₹", lowest)    
```
