```python
# Create a Student Marks List

marks = [78, 85, 90, 67, 88, 92, 76]

#1.print the marks list 
print("Student Marks List:", marks)
#2 Print the first element
print("First Mark:", marks[0])
#3Print the last element
print("Last Mark:", marks[-1])
#4 Print elements from index 2 to 5
print("Marks from index 2 to 5:", marks[2:6])
#5Find the number of elements
print("Number of Marks:", len(marks))
#6Find maximum marks
print("Maximum Marks:", max(marks))
#7Find minimum marks
print("Minimum Marks:", min(marks))
#8Find total marks
print("Total Marks:", sum(marks))
#9Sort marks in ascending order
print("Marks in Ascending Order:", sorted(marks))
#10 Sort marks in descending order
print("Marks in Descending Order:", sorted(marks, reverse=True))
#11 Add 95
print("Adding 95 to the marks list.")
marks.append(95)
print("Updated Marks List:", marks)
#12 Add [81, 84]
print("Adding [81, 84] to the marks list.")
marks.extend([81, 84])
print("Updated Marks List:", marks) 
#13 Remove 67
print("Removing 67 from the marks list.")
marks.remove(67)
print("Updated Marks List:", marks)
#14 Count how many times 90 occurs
print("Count of 90 in the marks list:", marks.count(90))    
#15 Find the index of 88
print("Index of 88 in the marks list:", marks.index(88))    


```
