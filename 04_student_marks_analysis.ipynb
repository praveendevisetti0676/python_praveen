```python
marks = [78, 92, 45, 67, 88, 53, 99]
#Analysis of marks to find 
# Count of 90+
# Count of 75-89
# Count of 50-74
# Count of below 50
count_90_plus = 0  
count_75_89 = 0
count_50_74 = 0 
count_below_50 = 0
for i in marks:
    if i >= 90: 
        count_90_plus += 1
    elif i >= 75: 
        count_75_89 += 1
    elif i >= 50: 
        count_50_74 += 1
    else:
        count_below_50 += 1
print("Count of 90+ marks: ", count_90_plus)
print("Count of 75-89 marks: ", count_75_89)    
print("Count of 50-74 marks: ", count_50_74)
print("Count of below 50 marks: ", count_below_50)
```
