```python
# Python Dictionary Challenge
student = {
    "name": "Rahul",
    "age": 22,
    "course": "Python",
    "city": "Bangalore",
    "marks": 88
}
#Print the complete dictionary
print("Student details:", student)
#Print the student's name
print("Student name:", student["name"])
#Print their course
print("Student course:", student["course"])
#Print all keys
print("All keys:", student.keys())
#Print all values
print("All values:", student.values())
#Print all key-value pairs
print("All key-value pairs:", student.items())
#Change marks from 88 to 92
student["marks"] = 92
print("Updated marks:", student["marks"])
#using update method to change age to 23 and city to "Mumbai"
student.update({"age": 23, "city": "Mumbai"})
print("Student details using update:", student)
#Add "email"
student["email"] = "rahul@example.com"
print("Student email:", student["email"])
#Add "phone"
student["phone"] = "1234567890"
print("Student phone:", student["phone"])
print("Updated student details:", student)
#Remove "city" pop ()
student.pop("city")
print("Student details after removing city:", student)
#Use get() to retrieve "name"
print("Student name (using get):", student.get("name"))
#Create a copy of the dictionary
student_copy = student.copy()
print("Copy of student details:", student_copy)
```
