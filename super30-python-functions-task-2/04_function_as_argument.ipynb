```python
#Pass a Function as an Argument 
def add(a, b):
    return a + b


def multiply(a, b):
    return a * b


def calculate(operation, a, b):
    return operation(a, b)


result1 = calculate(add, 10, 20)
result2 = calculate(multiply, 10, 20)

print("Addition:", result1)
print("Multiplication:", result2)
```
