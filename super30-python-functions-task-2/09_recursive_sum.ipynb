```python
#Recursive Sum from 1 to n 
def calculate_sum(number):
    if number == 0:
        return 0
    print("value in number is ", number)
    return number + calculate_sum(number - 1)


number = int(input("Enter a number: "))

if number < 0:
    print("Please enter a positive number.")
else:
    result = calculate_sum(number)
    print("Sum:", result)

```
