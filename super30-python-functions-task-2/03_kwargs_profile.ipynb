```python
#Create Dynamic Profile Using **kwargs 
#**kwargs allows a function to accept any number of keyword arguments.

def create_profile(**kwargs):
    print("--- User Profile ---")

    for key, value in kwargs.items():
        print(f"{key}: {value}")

create_profile(
    name="Rahul",
    city="Hyderabad",
    course="Python Programming",
    age=22
)
```
