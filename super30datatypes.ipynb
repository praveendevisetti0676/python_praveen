```python
# Python Data Type Laboratory
# PART 1: Create 15+ variables

# Integer variables
age = 25
marks = 95
year = 2026

# Float variables
height = 5.11
percentage = 97.5
salary = 75000.50

# String variables
name = "Praveen"
course = "Python Programming"
city = "Hyderabad"

# Boolean variables
likes_python = True
is_student = False

# List variable
student_marks = [78, 85, 90, 92]

# Tuple variable
coordinates = (17.3850, 78.4867)

# Set variable
unique_numbers = {10, 20, 30, 20}

# Dictionary variable
student = {
    "name": "Praveen",
    "age": 25,
    "course": "Python"
}

# PART 2: Print variables and their types

print("===== VARIABLES AND DATA TYPES =====")
print("Integer Variables:")
print("Age:", age, "| Type:", type(age))
print("Marks:", marks, "| Type:", type(marks))
print("Year:", year, "| Type:", type(year))
print("\nFloat Variables:")
print("Height:", height, "| Type:", type(height))
print("Percentage:", percentage, "| Type:", type(percentage))
print("Salary:", salary, "| Type:", type(salary))
print("\nString Variables:")
print("Name:", name, "| Type:", type(name))
print("Course:", course, "| Type:", type(course))
print("City:", city, "| Type:", type(city))
print("\nBoolean Variables:")
print("Likes Python:", likes_python, "| Type:", type(likes_python))
print("Is Student:", is_student, "| Type:", type(is_student))
print("\nList Variable:")
print("Student Marks:", student_marks, "| Type:", type(student_marks))
print("\nTuple Variable:")
print("Coordinates:", coordinates, "| Type:", type(coordinates))
print("\nSet Variable:")
print("Unique Numbers:", unique_numbers, "| Type:", type(unique_numbers))  
print("\nDictionary Variable:")
print("Student:", student, "| Type:", type(student))


# ==========================================
# PART 3: Basic Type Conversions
# ==========================================

print("\n===== TYPE CONVERSIONS =====")

# 1. String to Integer
value1 = int("100")
print("String to Integer:", value1, "| Type:", type(value1))

# 2. String to Float
value2 = float("99.99")
print("String to Float:", value2, "| Type:", type(value2))

# 3. Integer to String
value3 = str(123)
print("Integer to String:", value3, "| Type:", type(value3))

# 4. Float to Integer
value4 = int(99.99)
print("Float to Integer:", value4, "| Type:", type(value4))

# 5. Float to String
value5 = str(99.99)
print("Float to String:", value5, "| Type:", type(value5))

# 4. Integer to Boolean
value6 = bool(1)   
print("Integer to Boolean:", value6, "| Type:", type(value6))
```
