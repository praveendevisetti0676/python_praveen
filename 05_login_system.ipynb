```python
correct_password = "python123"
# Create a simple login system with a maximum of 3 password attempts.

login_attempts = 0
while login_attempts < 3:
    password = input("Enter your password: ")
    if password == correct_password:
        print("Login successful!")
        break
    else:
        login_attempts += 1
        print(f"Incorrect password. You have {3 - login_attempts} attempts left.")  
else:
    print("Your account has been locked due to too many incorrect attempts.")
```
