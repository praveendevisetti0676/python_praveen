```python
# Python Sets - Student Groups Challenge
python_students = {"Rahul", "Aman", "Priya", "Karan", "Neha"}
java_students = {"Priya", "Karan", "Rohit", "Simran"}
#1.Print both sets
print("Python Students:", python_students)
print("Java Students:", java_students)
#2 Find students learning either Python or Java
all_students = python_students.union(java_students)
print("Students learning either Python or Java:", all_students)
#3 Find students learning both
both_students = python_students.intersection(java_students) 
print("Students learning both Python and Java:", both_students)
#4 Find students learning only Python
only_python_students = python_students.difference(java_students)
print("Students learning only Python:", only_python_students)
#5 Find students learning only Java
only_java_students = java_students.difference(python_students)
print("Students learning only Java:", only_java_students)   
#6Find students belonging to exactly one group
exactly_one_group = python_students.symmetric_difference(java_students)
print("Students belonging to exactly one group:", exactly_one_group)
#7 Add a new student
new_student = "Anjali" 
python_students.add(new_student)
print("Updated Python Students after adding a new student:", python_students)
#8 Remove a student
student_to_remove = "Karan" 
python_students.discard(student_to_remove)  
print("Updated Python Students after removing a student:", python_students)
#9 Demonstrate discard
python_students.discard("Neha")
print("Python Students after attempting to remove a non-existent student:", python_students)
```
