```python
text = input("Enter a string: ")
# Check whether a string is a palindrome using loops.    
is_palindrome = True
for i in range(len(text) // 2):
    if text[i] != text[-i - 1]:
        is_palindrome = False
        break
if is_palindrome:
    print("The string is a palindrome.")
else:
    print("The string is not a palindrome.")

```
