```python
numbers = [10, 25, 8, 40, 15, 30]
# find the 2nd largest number in the list without using sort() or max()
largest = second_largest = numbers[0]
for n in numbers:
    if n > largest:
        second_largest = largest
        largest = n
    elif n > second_largest and n != largest:
        second_largest = n
print("The second largest number is:", second_largest)
print("The largest number is:", largest)
```
