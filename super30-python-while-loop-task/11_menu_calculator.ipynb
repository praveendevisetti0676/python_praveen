```python
#Menu Calculator
while True:

    print("\n--- Calculator ---")
    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print("5. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 5:
        print("Calculator closed.")
        break

    if choice >= 1 and choice <= 4:

        num1 = float(input("Enter first number: "))
        print("You entered1:", num1)
        num2 = float(input("Enter second number: "))
        print("You entered2:", num2)

        if choice == 1:
            print("Sum Result:", num1 + num2)

        elif choice == 2:
            print("Difference Result:", num1 - num2)

        elif choice == 3:
            print("Product Result:", num1 * num2)

        elif choice == 4:
            if num2 == 0:
                print("Cannot divide by zero.")
            else:
                print("Quotient Result:", num1 / num2)

    else:
        print("Invalid choice. Please try again.")
```
