```python
#Number Guessing Game 
myscret_number = 8
num = int(input("Guess the number between 1 and 10: "))
while num != myscret_number:
    if num < myscret_number:
        print("Too low! Try again.", num)
    else:
        print("Too high! Try again.", num)
    num = int(input("Guess the number between 1 and 10: "))
print("Congratulations! You guessed the number.",num)   
```
