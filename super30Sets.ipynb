```python
# Python Sets & Duplicate Data Challenge

numbers = [10, 20, 30, 20, 40, 10, 50, 30, 60]

#1 Print the original list
print("Original List:", numbers)

#2Convert the list into a set
unique_numbers = set(numbers)
print("Unique Numbers:", unique_numbers)
#3. Convert the set back into a list
unique_list = list(unique_numbers)
print("Unique Numbers as List:", unique_list)

#4 Print the number of original elements
print("Number of Original Elements:", len(numbers))

#5 Print the number of unique elements
print("Number of Unique Elements:", len(unique_numbers))

#6 Create another example using duplicate student names 
students = ["Alice", "Bob", "Charlie", "Alice", "David", "Bob", "Eve"]
unique_students = set(students)
print("Unique Students:", unique_students)
```
