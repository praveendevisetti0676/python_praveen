```python
# Shopping Cart Using a Python List
cart = ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"]

#1. Display all products
print("Products in the cart:", cart)

#2Access first and last products
print("First product:", cart[0])
print("Last product:", cart[-1])
#3 Add "Webcam"
cart.append("Webcam")
print("Products in the cart after adding Webcam:", cart)
#4 Insert "USB Hub" at index 2
cart.insert(2, "USB Hub")
print("Products in the cart after inserting USB Hub at index 2:", cart)
#5 Remove "Mouse"
cart.remove("Mouse")
print("Products in the cart after removing Mouse:", cart)
#6 Remove the last item using pop()
remove_item = cart.pop()
print("Removed item:", remove_item)
#7 Find the index of "Monitor"
product_index = cart.index("Monitor")
print("Index of Monitor:", product_index)
#8 Count occurrences of "Laptop"
laptop_count = cart.count("Laptop")
print("Occurrences of Laptop:", laptop_count)
#9 Create a copy of the cart
cart_copy = cart.copy()
print("Copy of the cart:", cart_copy)
#10 Reverse the cart
cart.reverse()
print("Reversed cart:", cart)
#11 Sort the products alphabetically
cart.sort(reverse=True)
print("Sorted cart:", cart)

```
