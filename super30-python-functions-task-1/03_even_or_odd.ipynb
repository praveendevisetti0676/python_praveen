```python
# Check Even or Odd using functions
def check_even_odd(num): 
    """Return whether a number is even or odd."""
    if num % 2 == 0:
        return "Even"
    else:
        return "Odd" 
num = int(input("Enter a number: "))
result = check_even_odd(num)    
print(f"The number {num} is {result}.")
```
