```python
# Function using print()
def add_using_print(a, b):
    """Print the sum of two numbers."""
    result = a + b
    print("Inside print function:", result)


# Function using return
def add_using_return(a, b):
    """Return the sum of two numbers."""
    return a + b


# Input values
num1 = 10
num2 = 20


# Demonstrating print()
print("--- Using print() ---")
add_using_print(num1, num2)


# Demonstrating return
print("\n--- Using return ---")
result = add_using_return(num1, num2)
print("Returned result:", result)


# Demonstrating why return is useful
new_result = result * 2
print("Result multiplied by 2:", new_result)
```
