```python
#Calculate Discount with Default Argument 
def calculate_discount(price, discount=10):
    discount_amount = price * discount / 100
    final_price = price - discount_amount
    return final_price


price = float(input("Enter price: "))

# Uses default discount of 10%
print("Price after 10% discount:", calculate_discount(price))

# Uses a custom discount
custom_discount = float(input("Enter custom discount percentage: "))
print(
    "Price after custom discount:",
    calculate_discount(price, custom_discount)
)
```
