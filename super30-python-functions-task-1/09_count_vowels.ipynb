```python
#Count Vowels Using Function 
def count_vowels(text):
    """Return the number of vowels in a string."""
    vowels = "aeiouAEIOU"
    count = 0

    for character in text:
        if character in vowels:
            count += 1

    return count


text = input("Enter a string: ")
print("Number of vowels:", count_vowels(text))
```
