```python
def student_profile(name, age, course):
    """Return a formatted student profile."""
    
    profile = (
        f"Name   : {name}\n"
        f"Age    : {age}\n"
        f"Course : {course}"
    )
    
    return profile


# Positional arguments
profile1 = student_profile("Praveen", 30, "Python")
print(profile1)


# Keyword arguments
profile2 = student_profile(
    name="Praveen",
    age=30,
    course="Python"
)
print(profile2)

student_profile("Praveen", 30, "Python")



student_profile(
    name="Praveen",
    age=30,
    course="Python"
}
```
