```python
message = " Welcome To Python Programming Class "
print("Original message:", message)

# 1.Remove extra spaces
print("After removing extra spaces:", message.strip())
#2. Convert everything to lowercase
print("After converting to lowercase:", message.lower())
#3. Convert everything to uppercase
print("After converting to uppercase:", message.upper())    
#4.Convert to title case
print("After converting to title case:", message.title())
#5Replace "Python" with "Advanced Python"
print("After replacing 'Python' with 'Advanced Python':", message.replace("Python", "Advanced Python"))
#6Check whether the string starts with "Welcome"
print("Does the message start with 'Welcome'?", message.startswith("Welcome"))
#7Check whether it ends with "Class"
print("Does the message end with 'Class'?", message.endswith("Class"))
#8Count occurrences of "o"
print("Number of occurrences of 'o':", message.count("o"))
#9 Find the position of "Programming"
print("Position of 'Programming':", message.find("Programming"))
#10 Split the sentence into words
print("After splitting the message into words:", message.split())  


```
