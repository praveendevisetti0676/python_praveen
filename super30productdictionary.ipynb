```python
# Python Dictionary - Laptop Product Challenge
laptop = {
    "brand": "Dell",
    "model": "XPS 15",
    "price": 120000,
    "ram": "16GB",
    "storage": "512GB SSD",
    "available": True
}
# Display the product
print("Laptop Product Details:", laptop)
#Print brand
print("Brand:", laptop["brand"])
#Print model
print("Model:", laptop["model"])
#Print price
print("Price:", laptop["price"])
#Change price
laptop["price"] = 12500
print("Updated Price:", laptop["price"])
#Add processor information
laptop["processor"] = "Intel Core Ultra i7"
print("Processor:", laptop["processor"])
print("Laptop Product Details:", laptop)
#Add GPU information
laptop["gpu"] = "NVIDIA GeForce RTX 3050"
print("GPU:", laptop["gpu"])
print("Laptop Product Details:", laptop)
#Change RAM to "32GB"
laptop["ram"] = "32GB"
print("Updated RAM:", laptop["ram"])
print("Laptop Product Details:", laptop)
# Remove "available"
del laptop["available"]
print("Laptop Product Details:", laptop)
#Print all keys
print("All Keys:", laptop.keys())
#print all values
print("All Values:", laptop.values())
#print all items
print("All Items:", laptop.items())

mobile = {
    "brand": "Apple",
    "model": "17 pro max",
    "price": 167000,
    "ram": "16GB",
    "storage": "512GB",
    "available": True
}

print("Mobile Product Details:", mobile)
#print all items
print("All Items:", mobile.items())
```
