```python
salaries = [45000, 55000, 60000, 400, 75000]


def total_payroll(salaries):
    total = 0

    for salary in salaries:
        total += salary

    return total


def average_salary(salaries):
    return total_payroll(salaries) / len(salaries)


def highest_salary(salaries):
    highest = salaries[0]

    for salary in salaries:
        if salary > highest:
            highest = salary

    return highest


def lowest_salary(salaries):
    lowest = salaries[0]

    for salary in salaries:
        if salary < lowest:
            lowest = salary

    return lowest


def above_average(salaries):
    average = average_salary(salaries)

    print("Employees earning above average:")

    for salary in salaries:
        if salary > average:
            print(salary)


print("Total Payroll:", total_payroll(salaries))
print("Average Salary:", average_salary(salaries))
print("Highest Salary:", highest_salary(salaries))
print("Lowest Salary:", lowest_salary(salaries))

above_average(salaries)
```
