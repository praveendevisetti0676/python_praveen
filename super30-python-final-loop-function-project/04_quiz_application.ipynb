```python
questions = [
    {
        "question": "What is the correct extension for Python files?",
        "options": ["A. .java", "B. .py", "C. .cpp", "D. .txt"],
        "answer": "B"
    },
    {
        "question": "Which keyword is used to define a function?",
        "options": ["A. function", "B. define", "C. def", "D. fun"],
        "answer": "C"
    },
    {
        "question": "Which loop is condition-controlled?",
        "options": ["A. while", "B. for", "C. both", "D. none"],
        "answer": "A"
    },
    {
        "question": "Which symbol is used for comments?",
        "options": ["A. //", "B. #", "C. /*", "D. --"],
        "answer": "B"
    },
    {
        "question": "Which data type stores key-value pairs?",
        "options": ["A. List", "B. Tuple", "C. Dictionary", "D. Set"],
        "answer": "C"
    }
]


def run_quiz():
    score = 0

    for item in questions:
        print("\n" + item["question"])

        for option in item["options"]:
            print(option)

        answer = input("Enter your answer: ").upper()

        if answer == item["answer"]:
            print("Correct!")
            score += 1
        else:
            print("Wrong!")

    percentage = (score / len(questions)) * 100

    print("\n--- QUIZ RESULT ---")
    print("Score:", score, "/", len(questions))
    print("Percentage:", percentage, "%")


run_quiz()
```
